```python
# Split the dataset into input features (X) and the target (y).
from sklearn.model_selection import train_test_split

X  = df.drop('target', axis = 1)
y = df['target']

# Keep 20% of the data for testing and 80% for training when spliting the data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.2, random_state = 1
)

print ('Training data size: ', X_train.shape)
print ('Testing data size: ', X_test.shape)
```

```python
# Check which class labels are present in the test set.
print (y_test.unique())

# Count how many examples of each class are in the training set
# to  check the target column are balanced.
from collections import Counter
Counter(y_train)
```

```python
# Standardize the features so that they are on a comparable scale
# This is especially useful for models that use gradient-based optimization
# and distance calculations
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# The training mean and standard deviation are learned
X_train = scaler.fit_transform(X_train)

# Transform the test data using the SAME scaling learned from X_train
X_test = scaler.transform(X_test)
```

```python
def sigmoid (z):
    # Convert the model's raw output into a value between 0 and 1
    # This value can be interpreted as the estimated probability of class 1
    return 1/(1+ np.exp(-z))

def train (X_train,y_train, learning_rate = 0.001, iteration =1000):

    # Create one weight of 0 for every input feature/column
    w = np.zeros(X_train.shape[1])

  
    b = 0

    for i in range(iteration):

        # Calculate the model's prediction:
        # z = Xw + b
        z = np.dot(X_train,w) + b

        # Convert the prediction into a probability
        prediction = sigmoid(z)

        # Calculate the gradient for each weight
        # X_train.T changes the shape so we get one gradient per feature
        dw = np.dot(X_train.T, (prediction - y_train)) /len(y_train)

        # Calculate the gradient for the bias.
        db = np.sum((prediction-y_train)) /len(y_train)

        # Update the weights and bias using gradient descent.
        # The gradient tells us which direction to move;
        # the learning rate controls how large the update is
        w = w - learning_rate*dw
        b = b - learning_rate*db

    return w, b
```

```python

w, b = train (X_train,y_train)

print ('Weight: ', w)
print ('Bias: ', b)
```

```python
# Use the learned weights and bias to calculate the output
# for the unseen test data
z = np.dot (X_test, w) + b

# Convert the output into probabilities using the sigmoid function
predictions = sigmoid(z)
```

```python
# Convert probabilities into class predictions.
# A probability >= 0.5 is classified as class 1;
# otherwise it is classified as class 0
predicted_classes = (predictions >= 0.5).astype(int)


print (predictions[0:10])


print (y_test[0:10])
```

```python

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Accuracy measures the percentage of test examples classified correctly
accuracy = accuracy_score(y_test, predicted_classes)
print (f"Accuracy: {round(accuracy*100, 2)}%")

# The confusion matrix shows how many examples were:
# true negatives, false positives, false negatives, and true positives
conf_matrix = confusion_matrix(y_test, predicted_classes)
print (f"Confusion matrix: ")
print (conf_matrix)

# The classification report provides precision, recall, F1-score,
# and support for each class
clas_report = classification_report (y_test, predicted_classes)
print (f"Classification report: ")
print (clas_report)
```

```python
# Count the number of examples in each class for the training set,
# test set, model predictions, and the complete dataset
# This helps us check whether the model is heavily favoring one class
print(Counter(y_train))
print(Counter(y_test))
print(Counter(predicted_classes))
print(Counter (df['target']))
```

```python
# Train a Gaussian Naive Bayes classifier as another model
# to compare against the logistic regression implementation
from sklearn.naive_bayes import GaussianNB

nb_model = GaussianNB()
nb_model.fit(X_train, y_train)

# Generate predictions for the unseen test data
nb_prediction = nb_model.predict(X_test)

# Evaluate the Naive Bayes predictions
accuracy = accuracy_score(y_test, nb_prediction)
print (f"Accuracy: {round(accuracy*100, 2)}%")

conf_matrix = confusion_matrix(y_test, nb_prediction)
print (f"Confusion matrix: ")
print (conf_matrix)

clas_report = classification_report (y_test, nb_prediction)
print (f"Classification report: ")
print (clas_report)
```

```python
# Train a K-Nearest Neighbors (KNN) classifier for comparison
from sklearn.neighbors import KNeighborsClassifier

# Use the 10 nearest neighbors when making a prediction
svc_model = KNeighborsClassifier(
    n_neighbors= 10
)

svc_model.fit(X_train, y_train)

# Generate predictions for the unseen test data
svc_prediction = svc_model.predict(X_test)

# Evaluate the KNN predictions using the same metrics
# used for the other models
accuracy = accuracy_score(y_test, svc_prediction)
print (f"Accuracy: {round(accuracy*100, 2)}%")

conf_matrix = confusion_matrix(y_test, svc_prediction)
print (f"Confusion matrix: ")
print (conf_matrix)

clas_report = classification_report (y_test, svc_prediction)
print (f"Classification report: ")
print (clas_report)
```
